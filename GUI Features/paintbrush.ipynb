Utilizing mouse as a paint brush

```python
import numpy as np 
import cv2

# mouse callback function for circle
# cv2. circle takes img, center point, radius, color, thickness
def draw_circle(event, x, y, flags, param):
    if event ==  cv2.EVENT_LBUTTONDBLCLK: # When left click on img
        cv2.circle(img, (x,y),100,(255,0,0),1)

img = np.zeros((512, 512, 3), np.uint8) # Black img screen
cv2.namedWindow('image')
cv2.setMouseCallback('image', draw_circle) # Calling the function 

while True:
    cv2.imshow('image', img)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cv2.destroyAllWindows()
```

For rectangle

```python
def draw_rectangle(event, x, y, flags, param):
    if event ==  cv2.EVENT_LBUTTONDBLCLK:
        cv2.rectangle(img, (x,y), (x+200, y+400), (0, 255, 0), 3)

img  = np.zeros((512, 512, 3), np.uint8) 
cv2.namedWindow('image')
cv2.setMouseCallback('image', draw_rectangle)

while True:
    cv2.imshow('image', img)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cv2.destroyAllWindows()
```
