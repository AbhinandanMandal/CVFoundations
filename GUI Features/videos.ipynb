Capture video from camera

```python
# cv2.VideoCapture() object helps to capture video using webcam or camera
# We need to pass 0, 1, -1 for configuration of cameras

import cv2
import numpy as np 

cap = cv2.VideoCapture(0) # Helps to capture video from webcam
while True:
    ret, frame = cap.read() # Capture frame-by-frame for the video
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY) # Captured frame convert into gray image

    # Displaying the resulting frame
    cv2.imshow('frame', gray)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# After breaking we've to release capture
cap.release()
cv2.destroyAllWindows() # Destroying all window
```

Playing with video file

```python
import numpy as np 
import cv2

cap = cv2.VideoCapture('videos/dogs.mp4') # Can give webcame or video file path
while cap.isOpened(): # While video is open, till then
    ret, frame = cap.read() # Read the video file

    # Resizing the frame
    frame = cv2.resize(frame, (640, 480))

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY) # Read the following frame of the video in gray color
    cv2.imshow('frame', gray) # Show the gray scaled video
    if cv2.waitKey(1) & 0xFF == ord('q'): # Quit condition
        break
cap.release()
cv2.destroyAllWindows()
```

Saving a video

```python
import numpy as np
import cv2

cap = cv2.VideoCapture(0) # Specifying webcame for video capture

# FourCC is a 4-byte code used to specify the video codec. Mostly XVID is preferable
fourcc = cv2.VideoWriter_fourcc(*'XVID') # Type of video it'll take 
out = cv2.VideoWriter('output.mp4', fourcc, 20.0, (640,480)) # VideoWriter take output file name, fourcc, fps, size

while cap.isOpened():
    ret,frame = cap.read() # Reading the frame
    if ret == True:
        frame = cv2.flip(frame, 1) # For every frame save it
        out.write(frame) # Saving it

        cv2.imshow('frame', frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

cap.release()
out.release()
cv2.destroyAllWindows()

```
