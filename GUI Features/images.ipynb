```python
import cv2 
import numpy as np 

# There are 3 fundamental ways in which we can read image
# cv2.IMREAD_COLOR, also can be represented by 1, loads a color image
# cv2.IMREAD_GRAYSCALE, also can be represented by 0, loads gray version of the image
# cv2.IMREAD_UNCHANGED, also by -1, loads image as such including alpha channel

# img = cv2.imread('images/virat.png', 0)
# img= cv2.imread("images/virat.png", 1)
img= cv2.imread("images/virat.png", -1)
# It determines the way we're reading the image
cv2.imshow('image', img) # For showing image
cv2.waitKey(0) # Values are in miliseconds for window automatic destroy
cv2.destroyAllWindows() # Close the window
```

`cv2.imwrite()` saves an image into .png format

```python
import numpy as np
import cv2

img= cv2.imread("images/virat.png", cv2.IMREAD_GRAYSCALE)
cv2.imshow("image", img)
k = cv2.waitKey(0)
if k == 'q' or 27: # q for quit, 27 for ESC button
    cv2.destroyAllWindows()
if k == ord('s'): # wait for 's' key to save and exit
    cv2.imwrite('virat-grey.png', img)
    cv2.destroyAllWindows()
```

```python
# We can also plot image using matplotlib
import matplotlib.pyplot as plt 
img = cv2.imread("images/virat.png", cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray', interpolation='bicubic')
plt.show()
```

```python
# For not having x, y axis
import matplotlib.pyplot as plt
img = cv2.imread("images/virat.png", cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray', interpolation='bicubic')
plt.xticks([]), plt.yticks([]) # Not having x, y axis
plt.show()
```
