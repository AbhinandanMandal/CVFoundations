```python
import numpy as np 
import cv2

# Creatign a black image
img = np.zeros((512, 512, 3), np.uint8)

# A line, (255, 0, 0) for blue colored line
cv2.line(img, (0,0), (511, 511), (255,0,0), 5)
cv2.rectangle(img, (384, 0), (510, 128), (0, 255, 0), 3) # Rectangle
cv2.circle(img, (447, 63), 63, (0, 0, 255), -1) # -1 for solid circle

# Writing into img
font  = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(img, 'OpenCV', (10, 500), font, 4, (255, 255, 255),2,cv2.LINE_AA) # cv2.LINE_AA for smooth text
cv2.imshow('drawings', img) # cv2 update original img, so showing img will work on it
cv2.waitKey(0)
cv2.destroyAllWindows()
```
