```python
import cv2
import matplotlib.pyplot as plt 

img = cv2.imread('home.jpg', cv2.IMREAD_COLOR_RGB) # Loading image
hist = cv2.calcHist([img],[0],None,[256],[0, 256]) # [0] for grayscale image
# print(hist)
plt.imshow(img),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.show()
plt.plot(hist)
plt.xlim([0, 256])
plt.xlabel('No Pixles')
plt.ylabel('Pixel Intensity')
plt.show()
```

```python
# we can make it for other  colours also
img = cv2.imread('home.jpg')  
color=('b','g','r')
for i, col in enumerate(color):
    hist = cv2.calcHist([img], [i],None,[256],[0, 256]) # [0] for grayscale image
    plt.plot(hist, color=col)
    plt.xlim([0, 256])
plt.show()
```

**For Masked Region**

```python
import numpy as np
img = cv2.imread('home.jpg') # original img
mask = np.zeros(img.shape[:2],np.uint8)
mask[100:300, 100:400] = 255
masked_img = cv2.bitwise_and(img, img, mask=mask) # Masked img

hist_full = cv2.calcHist([img],[0],None,[256],[0, 256])
hist_mask = cv2.calcHist([img],[0],mask,[256],[0,256])

plt.subplot(221), plt.imshow(img, 'gray')
plt.subplot(222), plt.imshow(mask, 'gray')
plt.subplot(223), plt.imshow(masked_img, 'gray')
plt.subplot(224), plt.plot(hist_full), plt.plot(hist_mask)
plt.xlim([0, 256])
plt.show()
```
