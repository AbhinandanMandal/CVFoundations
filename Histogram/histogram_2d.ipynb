```python
# In 1D historgam we're taking only colors, 
# In 2D we'll take hue and saturation

import cv2
import matplotlib.pyplot as plt 

img = cv2.imread('home.jpg')
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV) # For 2D, conversion into 2D is must
hist = cv2.calcHist([hsv],[0,1],None,[180,256],[0, 180, 0, 256])
plt.imshow(hist, interpolation='nearest') # For 2D showing
plt.title("2D HSV Histogram")
plt.xlabel("Saturation")
plt.ylabel("Hue")
plt.colorbar()
plt.show()
```
