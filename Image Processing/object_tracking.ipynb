```python
import cv2
import numpy as np

img = cv2.imread('bottles.png')
resized_img = cv2.resize(img, (640, 580))
# cv2.imshow('img', resized_img)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

# For masking, converting into HSV is much efficient
hsv = cv2.cvtColor(resized_img, cv2.COLOR_BGR2HSV)
# cv2.imshow('img', hsv)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

# For pink bottle, we need to define range for pink color for masking
lower_pink1 = np.array([0, 40, 100])
upper_pink1 = np.array([10, 255, 255])

lower_pink2 = np.array([170, 40, 100])
upper_pink2 = np.array([180, 255, 255])

mask1=cv2.inRange(hsv, lower_pink1, upper_pink1)
mask2 = cv2.inRange(hsv, lower_pink2, upper_pink2)

mask = mask1 + mask2 # Combining the mask for better result
# cv2.imshow('img', mask)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

# Only keeping the pink color by masking into original resized_img
res = cv2.bitwise_and(resized_img, resized_img, mask=mask)
cv2.imshow('img', res)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Masking for multiple colors

```python
import cv2
import numpy as np

img = cv2.imread('bottles.png')
resized_img = cv2.resize(img, (640, 580))
# cv2.imshow('img', resized_img)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

hsv = cv2.cvtColor(resized_img, cv2.COLOR_BGR2HSV)
# cv2.imshow('img', hsv)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

# For pink
lower_pink1 = np.array([0, 40, 100])
upper_pink1 = np.array([10, 255, 255])

lower_pink2 = np.array([170, 40, 100])
upper_pink2 = np.array([180, 255, 255])

pink_mask1 = cv2.inRange(hsv, lower_pink1, upper_pink1)
pink_mask2 = cv2.inRange(hsv, lower_pink2, upper_pink2)

pink_mask = pink_mask1+pink_mask2 # For pink mask

# For blue
lower_blue = np.array([95, 50, 50])
upper_blue = np.array([125, 255, 255])

blue_mask = cv2.inRange(hsv, lower_blue, upper_blue) # For blue mask

res_pink = cv2.bitwise_and(resized_img, resized_img, mask=pink_mask)
# cv2.imshow('res pink',res_pink)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
res_blue = cv2.bitwise_and(resized_img, resized_img, mask=blue_mask)
# cv2.imshow('res blue', res_blue)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

res = res_pink + res_blue # Masking both colors together
cv2.imshow('res', res)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
