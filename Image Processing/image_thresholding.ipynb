**Simple Thresholding**

If pixel value is greater than a threshold value, it is assigned one value (may be white), else it is assigned another value (may be black). The function used is `cv2.threshold`. 

```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt 

img = cv2.imread('BW.jpg', cv2.COLOR_BGR2GRAY) # Grayscaling image
resized_img = cv2.resize(img, (640, 480))
# cv2.imshow('img', resized_img)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

# For thresholding an image, cv2.threshold works better on grayscale image
# 127 is the thresh value
# If color gradient's color more than 127 then it's white else black
ret, thresh1 = cv2.threshold(resized_img, 127, 255, cv2.THRESH_BINARY) # Binary threshing, either black or white
cv2.imshow('img', thresh1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Binary Threshold Inverse
# Same concept like binary threhold, but in inverse way
ret, thresh2 = cv2.threshold(resized_img, 127, 255, cv2.THRESH_BINARY_INV)
cv2.imshow('img', thresh2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# For truncation, celling/cap of brightness
ret, thresh3 = cv2.threshold(resized_img, 127, 255, cv2.THRESH_TRUNC)
cv2.imshow('img', thresh3)
cv2.waitKey(0)
cv2.destroyAllWindows()

# If a pixel value is greater than the threshold, it gets chopped down (truncated) to equal the threshold value.
# If a pixel value is less than or equal to the threshold, it stays completely unchanged
```

```python
# TOZERO
ret, thresh4 = cv2.threshold(resized_img, 127, 255, cv2.THRESH_TOZERO)
cv2.imshow('img', thresh4)
cv2.waitKey(0)
cv2.destroyAllWindows()

# cv2.THRESH_TOZERO completely erases dark areas while leaving bright areas untouched. 
# It modifies your image based on this rule:If a pixel value is greater than the threshold, it stays completely unchanged.
# If a pixel value is less than or equal to the threshold, it gets set to 0 (pure black).
```

```python
# Similarly TOZERO_INV
ret, thresh5 = cv2.threshold(resized_img, 127, 255, cv2.THRESH_TOZERO_INV)
cv2.imshow('img', thresh5)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
virat_img = cv2.imread('virat.png', cv2.COLOR_BGR2GRAY)
ret, thresh6 = cv2.threshold(virat_img, 127, 255, cv2.THRESH_BINARY)
ret, thresh7 = cv2.threshold(virat_img, 127, 255, cv2.THRESH_BINARY_INV)
ret, thresh8 = cv2.threshold(virat_img, 127, 255, cv2.THRESH_TRUNC)
ret, thresh9 = cv2.threshold(virat_img, 127, 255, cv2.THRESH_TOZERO)
ret, thresh10 = cv2.threshold(virat_img, 127, 255, cv2.THRESH_TOZERO_INV)


titles = ['Original Image', 'BINARY',
          'BINARY_INV', 'TRUNC', 'TOZERO', 'TOZERO_INV']
images = [virat_img, thresh6, thresh7, thresh8, thresh9, thresh10]

for i in range(6):
    plt.subplot(2, 3, i+1), plt.imshow(images[i], 'gray')
    plt.title(titles[i])
    plt.xticks([]), plt.yticks([])

plt.show()
```

**Adaptive Thresholding**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 

img = cv2.imread('puzzle.png')
# cv2.imshow('img', img)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

# Converting into gray for thresholding works
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Simple Thresholding
ret, thresh1 = cv2.threshold(gray_img, 127, 255, cv2.THRESH_BINARY)
# cv2.imshow('img', thresh1)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

# Adaptive thresholding
# Adaptive thresh mean
thresh2 = cv2.adaptiveThreshold(gray_img, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 2)
# cv2.imshow('img', thresh2)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

# Adaptive thresh gaussian
thresh2 = cv2.adaptiveThreshold(
    gray_img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)
cv2.imshow('img', thresh2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
