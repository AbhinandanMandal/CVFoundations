**Erosion**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('j.png')
# cv2.imshow('img',img)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
plt.imshow(img)
plt.title('img')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
# In erosion, kernel slides through the image, and erode the boundaries of foreground object
kernel = np.ones((5,5),np.uint8)
erosion  = cv2.erode(img, kernel, iterations=1) # Erosion for 1 iteration
plt.imshow(erosion)
plt.title('img')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
# for multiple iterations
erosion2 = cv2.erode(img, kernel, iterations=2) 
erosion5 = cv2.erode(img, kernel, iterations=5) 
erosion10 = cv2.erode(img, kernel, iterations=10) 
plt.subplot(131), plt.imshow(erosion2), plt.title('Erosion 2')
plt.xticks([]), plt.yticks([])
plt.subplot(132), plt.imshow(erosion5), plt.title('Erosion 5')
plt.xticks([]), plt.yticks([])
plt.subplot(133), plt.imshow(erosion10), plt.title('Erosion 10')
plt.xticks([]), plt.yticks([])
```

**Dilation**

```python
# For dialating image
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('j.png') # Original image
kernel = np.ones((5, 5),np.uint8)

dilate1 = cv2.dilate(img, kernel, iterations=1)
dilate2 = cv2.dilate(img, kernel, iterations=2)
dilate3 = cv2.dilate(img, kernel, iterations=3)

plt.subplot(141), plt.imshow(img), plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(142), plt.imshow(dilate1), plt.title('Dilate 1')
plt.xticks([]), plt.yticks([])
plt.subplot(143), plt.imshow(dilate2), plt.title('Dilate 2')
plt.xticks([]), plt.yticks([])
plt.subplot(144), plt.imshow(dilate3), plt.title('Dilate 3')
plt.xticks([]), plt.yticks([])
```

**Opening**

```python
# Opening is another name of EROSION FOLLOWED BY DILATION : (A⊖B)⊕B
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('j.png') # Original image
kernel = np.ones((5, 5), np.uint8)
opening = cv2.morphologyEx(img, cv2.MORPH_CLOSE,kernel)

plt.subplot(121), plt.imshow(img), plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(opening), plt.title('Opening')
plt.xticks([]), plt.yticks([])
```

**Closing**

```python
# Closing is DILATION FOLLOWED BY EROSION : (A⊕B)⊖B
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('j.png')  # Original image
kernel = np.ones((5, 5), np.uint8)
closing = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)

plt.subplot(121), plt.imshow(img), plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(opening), plt.title('Closing')
plt.xticks([]), plt.yticks([])
```

**Morphological Gradient**

```python
# It's difference between Dilation and Erosion of an image
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('j.png')  # Original image
kernel = np.ones((5, 5), np.uint8)
gradient = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, kernel)

plt.subplot(121), plt.imshow(img), plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(gradient), plt.title('Gradient')
plt.xticks([]), plt.yticks([])
```

**Top Hat**

```python
# Top Hat : It's the difference between input image and opening of an input image
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('j.png')  # Original image
kernel = np.ones((5, 5), np.uint8)
tophat = cv2.morphologyEx(img, cv2.MORPH_TOPHAT, kernel)

plt.subplot(121), plt.imshow(img), plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(tophat), plt.title('Tophat')
plt.xticks([]), plt.yticks([])
```
