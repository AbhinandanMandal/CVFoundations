OpenCV provides 3 types of high-pass filters, **Sobel, Scharr, Laplacian derivative**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 

img = cv2.imread('puzzle.png')
plt.imshow(img), plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
sobelx = cv2.Sobel(img, cv2.CV_64F,1,0, ksize=5) # cv2.CV_64F means desired image output depth, current is 64 bit float
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=5)
laplacian = cv2.Laplacian(img, cv2.CV_64F)

plt.subplot(131), plt.imshow(laplacian), plt.title('Laplacian')
plt.xticks([]), plt.yticks([])

plt.subplot(132), plt.imshow(sobelx), plt.title('Sobel x')
plt.xticks([]), plt.yticks([])

plt.subplot(133), plt.imshow(sobely), plt.title('Sobel y')
plt.xticks([]), plt.yticks([])
```
