Two images were added with formula `g(x) = (1-alpha).f_0(x) + alpha.f_1(x) + gamma`

By varying the values of `alpha`, `gamma` we can make image blending

```python
import cv2
import numpy as np

img1 = cv2.imread('opencv-logo.png', cv2.IMREAD_COLOR)
img2 = cv2.imread('robot.jpg', cv2.IMREAD_COLOR)
resized_img2 = cv2.resize(img2, (180, 222)) # For blending two images size has to be same

# Blending of image
dst = cv2.addWeighted(img1, 0.7, resized_img2, 0.3, 0)
cv2.imshow('dst', dst)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Bitwise Operations

```python
import cv2
import numpy as np

img1 = cv2.imread('virat.png')
img2 = cv2.imread('opencv-logo.png')

# For putting logo on the top-left corner, creating a ROI
# We need to put the the image, so we need to take height, width of the image
rows, cols, channels = img2.shape 
# And we need to crate a ROI of the same height, width onto the img1
roi = img1[0:rows, 0:cols]

# Creating a mask of the logo
img2gray = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY) # Image into gray scale
ret, mask = cv2.threshold(img2gray, 10, 255, cv2.THRESH_BINARY) # Mask of the image
mask_inv = cv2.bitwise_not(mask) # Inversing mask

# Blackout the area of the logo into ROI
img1_bg = cv2.bitwise_and(roi, roi, mask= mask_inv)
# Taking only the logo image area
img2_fg = cv2.bitwise_and(img2, img2, mask=mask)
# Putting the logo in ROI and modify the main image
dst = cv2.add(img1_bg, img2_fg)
img1[0:rows, 0:cols] = dst

cv2.imshow('res', img1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
