```python
# To make border around the image, cv2.copyMakeBorder() function is useful
import cv2
import numpy as np 
import matplotlib.pyplot as plt 

img = cv2.imread("opencv-logo.png", cv2.IMREAD_COLOR)

replicate = cv2.copyMakeBorder(img, 10, 10, 10, 10, cv2.BORDER_REPLICATE) # OpenCV loads an image into BGR format
# While matplotlib expects in RGB
# So, we need to change it back into RGB for plotting
replicate_rgb  = cv2.cvtColor(replicate, cv2.COLOR_BGR2RGB)
plt.imshow(replicate_rgb)
plt.title('REPLICATE')
plt.axis("off")
plt.show()

# cv2.BORDER_REPLICATE: Last element is replicated throughoutput
```

```python
# Reflect, for reflection
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("opencv-logo.png", cv2.IMREAD_COLOR)
reflect = cv2.copyMakeBorder(img, 10, 10, 10, 10, cv2.BORDER_REFLECT)
reflect_rgb = cv2.cvtColor(reflect, cv2.COLOR_BGR2RGB)
plt.imshow(reflect_rgb)
plt.title("REFLECT")
plt.axis('off')
plt.show()
```

```python
# BORDER REFLECT 101, similar to previous
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("opencv-logo.png", cv2.IMREAD_COLOR)
reflect_101 = cv2.copyMakeBorder(img, 10, 10, 10, 10, cv2.BORDER_REFLECT_101)
reflect_101_rgb = cv2.cvtColor(reflect_101, cv2.COLOR_BGR2RGB)
plt.imshow(reflect_101_rgb)
plt.title("REFLECT 101")
plt.axis('off')
plt.show()
```

```python
# Constant border color
import cv2
import numpy as np
import matplotlib.pyplot as plt
BLUE = [255, 0, 0] # For blue border color

img = cv2.imread("opencv-logo.png", cv2.IMREAD_COLOR)
constant = cv2.copyMakeBorder(img, 10, 10, 10, 10, cv2.BORDER_CONSTANT, value=BLUE)
constant_rgb = cv2.cvtColor(constant, cv2.COLOR_BGR2RGB)
plt.imshow(constant_rgb)
plt.title("CONSTANT BORDER")
plt.axis('off')
plt.show()
```
