```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 

img = cv2.imread('messi.png')
mask = np.zeros(img.shape[:2], np.uint8)

bgdModel = np.zeros((1,65), np.float64)
fgdModel = np.zeros((1,65), np.float64)

rect = (50, 50, 600, 600) # creating a rect for the foreground extraction
cv2.grabCut(img, mask, rect, bgdModel, fgdModel, 4, cv2.GC_INIT_WITH_RECT)

mask2 = np.where((mask==2)|(mask==0), 0, 1).astype('uint8')
gc_img = img*mask2[:,:,np.newaxis]

plt.imshow(img), plt.title('Original'), plt.xticks([]), plt.yticks([])
plt.show()
plt.imshow(gc_img), plt.title('Grab Cut Image'), plt.xticks([]), plt.yticks([])
plt.show()
```
