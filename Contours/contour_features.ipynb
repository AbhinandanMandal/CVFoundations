```python
# moment describe shape, size, position, orientation
import cv2
import numpy as np
import matplotlib.pyplot as plt 

img = cv2.imread('star.png', cv2.IMREAD_GRAYSCALE)
ret, thresh = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
cnt = contours[0]
M = cv2.moments(cnt) # moments
print(M)
```

```python
# contour area
area = cv2.contourArea(cnt)
print(area) # area and others will be on pixels
perimeter = cv2.arcLength(cnt, True)
print(perimeter)
```

```python
# Contour approximation
img = cv2.imread('test2.png', cv2.IMREAD_COLOR)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
ret, thresh = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE,cv2.CHAIN_APPROX_SIMPLE)
cnt = contours[0]

epsilon = 0.1*cv2.arcLength(cnt, True)
approx = cv2.approxPolyDP(cnt, epsilon, True)

for data in contours:
    print(f"Contours have this data {data}")
cont = cv2.drawContours(img, contours, -1, (0, 255, 0), 3)
plt.imshow(cont), plt.title('Contour')
plt.xticks([]), plt.yticks([])
plt.show()
```
