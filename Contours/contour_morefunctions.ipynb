```python
# Convexity defects
# we can also visualize that defect
import cv2
import numpy as np 
import matplotlib.pyplot as plt 

img = cv2.imread('star.png')
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
ret, thresh = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
cnt = contours[0]

haull = cv2.convexHull(cnt, returnPoints=False)
defects = cv2.convexityDefects(cnt, haull)

for i in range(defects.shape[0]):
    s, e, f, d = defects[i, 0]
    start = tuple(cnt[s][0])
    end = tuple(cnt[e][0])
    far = tuple(cnt[f][0])
    cv2.line(img, start, end, [0, 255, 0], 2)
    cv2.circle(img, far, 5, [0, 0, 255], -1)

cv2.imshow('img', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# For finding the shortest distance between a point in the image and a contour
dist = cv2.pointPolygonTest(cnt, (50,50), True)
print(dist)
```

```python
# To check if the two shapes are matching or not
img1 = cv2.imread('star.png')
img1_gray = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
img2 = cv2.imread('star2.png')
img2_gray = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

ret, thresh1 = cv2.threshold(img1_gray, 127, 255, cv2.THRESH_BINARY)
ret, thresh2 = cv2.threshold(img2_gray, 127, 255, cv2.THRESH_BINARY)

contours1, hierarchy = cv2.findContours(thresh1, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
contours2, hierarchy = cv2.findContours(thresh2, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
cnt1 = contours1[0]
cnt2 = contours2[0]

ret = cv2.matchShapes(cnt1, cnt2, 1, 0.0)
print(ret)
```
