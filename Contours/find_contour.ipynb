```python
# Contours can be explained as, curve joining all the continuous points
# having same color or intensity

import cv2
import numpy as np 
import matplotlib.pyplot as plt 

img = cv2.imread('test1.png', cv2.IMREAD_COLOR)
img_gray  =cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) # grayscaling
# binary formation
ret, thresh = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)
# finding contour points
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

for data in contours:
    print(f"Contours have this data {data}")
cont = cv2.drawContours(img, contours, -1, (0, 255, 0), 3)
plt.imshow(cont), plt.title('Contour')
plt.xticks([]), plt.yticks([])
plt.show()
```
